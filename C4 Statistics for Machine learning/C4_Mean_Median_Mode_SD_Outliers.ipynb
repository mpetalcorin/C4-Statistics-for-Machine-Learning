```python
import numpy as np

weights = [8319, 320, 12000, 4489, 12000, 13449, 10398, 7266, 10922, 6267, 9034, 8449, 12308, 4959, 12280, 9353, 8470, 6353, 69420, 10450, 12385, 5000, 12117, 10308, 5000, 5010, 6345, 10723, 23, 6873, 13926, 12713, 11718, 9967, 6675, 5000, 6085, 11560, 8635, 11422, 13809, 11819, 12005, 5804, 9751, 12416, 29232, 4411, 7726, 7300]

# Create an array of indices from 1 to 50
indices = np.arange(1, 51)

# Combine weights and indices into a 2D array
data = np.column_stack((indices, weights))

print(data)
```

```python
# Calculate percentiles based on the weights
lower_percentile = np.percentile(weights, 4)
upper_percentile = np.percentile(weights, 96)

# Identify outliers based on weight
outliers_weight = data[np.logical_or(data[:, 1] < lower_percentile, data[:, 1] > upper_percentile)]

print("Outliers based on weight:", outliers_weight)
```
